# 03 - Inference with the saved model

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))  # repo root
from heartml.predict import load_model, predict_records
from heartml.schema import SAMPLE_PAYLOAD, validate_record
model = load_model()
rec, errs = validate_record(SAMPLE_PAYLOAD)
assert not errs
predict_records(model, [rec])
```

```python
import json, urllib.request
# against a running API (docker run -p 8000:8000 heart-disease-api:1.0.0)
req = urllib.request.Request('http://localhost:8000/predict', json.dumps(SAMPLE_PAYLOAD).encode(), {'Content-Type': 'application/json'})
print(urllib.request.urlopen(req).read().decode())
```
